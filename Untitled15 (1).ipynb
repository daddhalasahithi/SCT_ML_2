```python
# SkillCraft Technology - Machine Learning Internship
# Task 02: K-Means Clustering for Customer Segmentation
# Track Code: ML (Dataset: Mall_Customers.csv)

# 1. Import Required Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans

print("✅ Libraries successfully imported!")

# 2. Load the Dataset
file_name = 'Mall_Customers.csv'
df = pd.read_csv(file_name)

print("\n--- Dataset Preview (First 5 Rows) ---")
print(df.head())

# 3. Select Features for Clustering
# Extracting 'Annual Income (k$)' and 'Spending Score (1-100)' columns (Columns at index 3 and 4)
X = df.iloc[:, [3, 4]].values

print(f"\n✅ Features selected. Data shape for clustering: {X.shape}")

# 4. Use the Elbow Method to Find the Optimal Number of Clusters (K)
wcss = []  # Within-Cluster Sum of Squares
for i in range(1, 11):
    kmeans = KMeans(n_clusters=i, init='k-means++', random_state=42, n_init=10)
    kmeans.fit(X)
    wcss.append(kmeans.inertia_)

# Plotting the Elbow Graph
plt.figure(figsize=(8, 5))
plt.plot(range(1, 11), wcss, marker='o', linestyle='--', color='purple', linewidth=2)
plt.title('The Elbow Method to Find Optimal K')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('WCSS (Inertia)')
plt.grid(True)
plt.show()

print("\n💡 Observation: Look at the Elbow Graph above. The 'elbow point' where the drop sharply flattens out is at K=5.")

# 5. Train the K-Means Model with the Optimal Number of Clusters (K = 5)
optimal_k = 5
kmeans = KMeans(n_clusters=optimal_k, init='k-means++', random_state=42, n_init=10)
y_kmeans = kmeans.fit_predict(X)

# 6. Visualizing the Customer Clusters
plt.figure(figsize=(10, 7))

# Plotting the clusters with group profiling labels
colors = ['teal', 'orange', 'green', 'red', 'magenta']
cluster_labels = [
    'Low Income, High Spend',
    'High Income, High Spend (Target Group)',
    'Low Income, Low Spend',
    'High Income, Low Spend',
    'Average Income, Average Spend'
]

for i in range(optimal_k):
    plt.scatter(X[y_kmeans == i, 0], X[y_kmeans == i, 1], s=100, c=colors[i], label=cluster_labels[i], alpha=0.8)

# Plotting the cluster centers (Centroids)
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], s=250, c='black', marker='X', label='Centroids')

plt.title('Customer Segments (K-Means Clustering - SCT_ML_2)')
plt.xlabel('Annual Income (k$)')
plt.ylabel('Spending Score (1-100)')
plt.legend(loc='best')
plt.grid(True)
plt.show()

print("✅ Customer Segments and Centroids mapped successfully!")
```
